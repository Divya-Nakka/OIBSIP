# Age — Median Imputation

### Missing Age

Age is a numerical variable. Median imputation is selected because age
may contain extreme values, and the median is less affected by outliers
than the mean.

Therefore, missing Age values will be replaced with the median Age.

```python
age_median = df_messy['Age'].median()

df_messy['Age'] = df_messy['Age'].fillna(age_median)

print("Missing Age values:", df_messy['Age'].isnull().sum())
```

# Handle Missing Embarked

### Missing Embarked

Embarked is a categorical variable. Since only a small number of values
are missing, mode imputation is appropriate.

The most frequently occurring embarkation location will be used.

```python
embarked_mode = df_messy['Embarked'].mode()[0]

df_messy['Embarked'] = df_messy['Embarked'].fillna(embarked_mode)

print("Missing Embarked values:", df_messy['Embarked'].isnull().sum())
```

# Handle Missing Fare

### Missing Fare

Fare is a numerical variable. Median imputation is used because ticket
fares contain extreme values, making the median more robust than the mean.

```python
fare_median = df_messy['Fare'].median()

df_messy['Fare'] = df_messy['Fare'].fillna(fare_median)

print("Missing Fare values:", df_messy['Fare'].isnull().sum())
```

# Handle Cabin

### Missing Cabin

The Cabin column contains a large number of missing values. Removing
all rows with missing Cabin information would result in substantial
data loss.

Therefore, a new binary feature `CabinKnown` will be created to indicate
whether cabin information is available. The original Cabin column will
then be removed because it is not required for this cleaning exercise.

```python
df_messy['CabinKnown'] = df_messy['Cabin'].notnull().astype(int)

df_messy.drop(columns=['Cabin'], inplace=True)

df_messy.head()
```

# Remove Duplicate Rows

```python
print("Duplicate rows before removal:", df_messy.duplicated().sum())
```

```python
duplicates_removed = df_messy.duplicated().sum()

df_messy = df_messy.drop_duplicates().reset_index(drop=True)

print("Duplicates removed:", duplicates_removed)
print("Dataset shape after removal:", df_messy.shape)
```

### Duplicate Removal

Exact duplicate rows were removed because they represent repeated records
and could cause bias during analysis.

Only exact duplicate records were removed. Similar-looking but potentially
valid passenger records were retained.

# Standardise Gender

```python
df_messy['Sex'].value_counts(dropna=False)
```

```python
df_messy['Sex'] = (
    df_messy['Sex']
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({
        'm': 'male',
        'male': 'male',
        'f': 'female',
        'female': 'female'
    })
)

df_messy['Sex'].value_counts()
```

```python
df_messy['Sex'] = df_messy['Sex'].str.title()

df_messy['Sex'].value_counts()
```

### Gender Standardisation

Gender values contained inconsistent capitalization, abbreviations,
and whitespace.

Values such as `M`, `male`, and `MALE` were standardised to `Male`,
while `F`, `female`, and `FEMALE` were standardised to `Female`.

# Standardise Embarked

```python
df_messy['Embarked'] = (
    df_messy['Embarked']
    .astype(str)
    .str.strip()
    .str.upper()
)
```

```python
df_messy['Embarked'].value_counts()
```

# Correct Data Types

```python
df_messy.dtypes
```

```python
df_messy['PassengerId'] = df_messy['PassengerId'].astype(str)

df_messy['Survived'] = df_messy['Survived'].astype(int)

df_messy['Pclass'] = df_messy['Pclass'].astype(int)

df_messy['Age'] = pd.to_numeric(df_messy['Age'], errors='coerce')

df_messy['Fare'] = pd.to_numeric(df_messy['Fare'], errors='coerce')

df_messy['SibSp'] = df_messy['SibSp'].astype(int)

df_messy['Parch'] = df_messy['Parch'].astype(int)

df_messy['CabinKnown'] = df_messy['CabinKnown'].astype(int)
```

```python
df_messy.dtypes
```

# Detect Outliers Using IQR

```python
numeric_columns = ['Age', 'Fare', 'SibSp', 'Parch']

outlier_report = []

for column in numeric_columns:
    Q1 = df_messy[column].quantile(0.25)
    Q3 = df_messy[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_messy[
        (df_messy[column] < lower_bound) |
        (df_messy[column] > upper_bound)
    ]

    outlier_report.append({
        'Column': column,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers)
    })

outlier_report = pd.DataFrame(outlier_report)

outlier_report
```

# Investigate the Fare Outlier

```python
df_messy[df_messy['Fare'] > 1000]
```

### Fare Outlier Treatment

The Fare column contains an extreme value of 9999, which is inconsistent
with the normal fare distribution and was identified as an artificial
data-entry anomaly.

Because this value is considered invalid rather than a legitimate extreme
observation, it will be replaced with the median Fare value.

```python
fare_median = df_messy['Fare'].median()

df_messy.loc[df_messy['Fare'] > 1000, 'Fare'] = fare_median
```
